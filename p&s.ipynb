```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.stats import spearmanr

# Sample Data
study_hours = [1, 2, 3, 4, 5]
exam_scores = [40, 50, 60, 70, 80]

# Calculate Spearman Correlation
result = spearmanr(study_hours, exam_scores)

print("Spearman Correlation:", result.statistic)
print("P-value:", result.pvalue)

# --- Visualization ---
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 1. Scatter Plot with Monotonic Trend Line
sns.regplot(
    x=study_hours,
    y=exam_scores,
    ax=axes[0],
    color="teal",
    marker="o",
    scatter_kws={"s": 70},
)
axes[0].set_title(
    f"Scatter Plot\n(Spearman ρ = {result.statistic:.2f})", fontsize=12
)
axes[0].set_xlabel("Study Hours")
axes[0].set_ylabel("Exam Scores")
axes[0].grid(True, linestyle="--", alpha=0.6)

# 2. Heatmap Representation
data = pd.DataFrame({"Study Hours": study_hours, "Exam Scores": exam_scores})
corr_matrix = data.corr(method="spearman")

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f",
    ax=axes[1],
    cbar=True,
)
axes[1].set_title("Spearman Correlation Heatmap", fontsize=12)

plt.tight_layout()
plt.savefig("spearman")
plt.show()
```

```python
from scipy.stats import pearsonr, spearmanr

study_hours = [1, 2, 3, 4, 5]
marks = [10, 20, 35, 60, 100]

print("Pearson:", pearsonr(study_hours, marks).statistic)
print("Spearman:", spearmanr(study_hours, marks).statistic)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, spearmanr

# Data
study_hours = [1, 2, 3, 4, 5]
marks = [10, 20, 35, 60, 100]

# Calculations
p_corr, _ = pearsonr(study_hours, marks)
s_corr, _ = spearmanr(study_hours, marks)

print(f"Pearson: {p_corr:.4f}")
print(f"Spearman: {s_corr:.4f}")

# --- Matplotlib & Seaborn Visualization ---
plt.figure(figsize=(8, 5))

# Scatter plot and trendline
sns.regplot(
    x=study_hours,
    y=marks,
    color="darkblue",
    marker="o",
    scatter_kws={"s": 80},
    line_kws={"linestyle": "--", "color": "red"},
)

# Titles and Labels
plt.title("Study Hours vs Marks", fontsize=14, fontweight="bold")
plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Marks", fontsize=12)

# Graph paina Pearson & Spearman values text ga add cheyadaniki:
plt.text(
    1.2,
    85,
    f"Pearson Correlation: {p_corr:.2f}\nSpearman Correlation: {s_corr:.2f}",
    fontsize=11,
    bbox=dict(
        boxstyle="round,pad=0.5", facecolor="lightyellow", edgecolor="gray"
    ),
)
# plt.legend("x_label","Y_label")

plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.savefig("p&s")
plt.show()
```
