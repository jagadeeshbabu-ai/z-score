```python
import numpy as np

```

```python
data=np.array([10, 12, 11, 13, 12, 14, 100])

mean=np.mean(data)
std=np.std(data)
```

```python
z_score=(data-mean)/std
print("z_socre:",z_score)
print("Mean:",mean)
print("Standivation:",std)

outlier=data[np.abs(z_score)>2]
print("Outliers:",outlier)
```

pandas

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "salary": [25000, 27000, 26000, 28000, 30000, 29000, 200000]
})

mean = df["salary"].mean()
std = df["salary"].std()

df["z_score"] = (df["salary"] - mean) / std

# Detect outliers
outliers = df[np.abs(df["z_score"]) > 3]

print(df)
print("\nOutliers:")
print(outliers)
```

scipy

```python
from scipy.stats import zscore
import pandas as pd

df = pd.DataFrame({
    "salary": [25000, 27000, 26000, 28000, 30000, 29000, 200000]
})

df["z_score"] = zscore(df["salary"])

outliers = df[abs(df["z_score"]) > 3]

print(outliers)
# z = (x - mean) / standard_deviation
```
